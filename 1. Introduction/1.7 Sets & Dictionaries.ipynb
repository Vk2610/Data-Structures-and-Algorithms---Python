# Sets and Dictionaries

## 1 Sets :

In Python, a set is a built-in data type that stores an unordered collection of unique and immutable elements. While the set itself is mutable (you can add or remove items), the individual items inside it must be of a hashable type, such as strings, integers, or tuples.

```python
# Sets example 
my_set = {1, 2, 3, 4, 5}
print(my_set)  
```

```python
# Operations/Functions of Sets
# 1. Add an element to a set
my_set.add(6)
print(my_set)
```

```python
# 2 . Remove an element from a set
my_set.remove(6)
print(my_set)
```

```python
# 3. Check if an element is in a set
print(3 in my_set)  # Output: True
print(6 in my_set)  # Output: False
```

```python
# 4. Length of a set
print(len(my_set))  # Output: 5
```

```python
# Union of set
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}

print(set1.union(set2))
```

```python
# Intersection of set
print(set1.intersection(set2))
```

```python
# Deep and shallow copy of set
# Shallow copy
set1 = {1, 2, 3, 4, 5}
set2 = set1.copy()
print(set2)


# Note: do not use variable name 'set', because it shadows Python's built-in set() constructor
```

```python
# Difference
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}
print(set2-set1)
print(set2.difference(set1))
```

```python
# symmetric difference
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}
print(set1^set2)
print(set1.symmetric_difference(set2))
```

```python
# DSA Problems on SETS

# Q1. Given two lists, write a Python program to find the common elements between them using sets.
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

common_elements = set(list1) & set(list2)
print(common_elements)
```

```python
# Q2. Write a Python program to remove duplicates from a list using sets.
my_list = [1, 2, 3, 4, 5, 1, 2, 3]
unique_elements = list(set(my_list))
print(unique_elements)
```

```python
# Q3. Write a Python program to find the difference between two lists using sets.
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

difference = set(list1) - set(list2)
print(difference)
```
